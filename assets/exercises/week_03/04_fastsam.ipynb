## 2. FastSAM mit Text-Prompting  
  
FastSAM ermöglicht schnelle Bildsegmentierung basierend auf Text-Prompts (z. B. "bus").

```python
from ultralytics import FastSAM
from PIL import Image

# Bild-Quelle von der Ultralytics-Webseite
source = "https://ultralytics.com/images/bus.jpg"

# FastSAM-Modell laden
model = FastSAM("FastSAM-s.pt")  # oder FastSAM-x.pt

# Inferenz auf dem Bild mit Text-Prompt "bus" ausführen
everything_results = model(
    source,
    device="cpu",
    retina_masks=True,
    imgsz=1024,
    conf=0.4,
    iou=0.9,
    texts="bus"
)

# Ergebnisse visualisieren
for r in everything_results:
    im_array = r.plot()  # BGR Numpy-Array
    im = Image.fromarray(im_array[..., ::-1])  # RGB PIL-Bild
    display(im)
```

## Zusammenfassung  
  
- **FastSAM**: Erstellt präzise Segmentierungsmasken in Echtzeit.  
- **Text Prompting**: Filtert Segmentierungsmasken anhand von Text-Prompts (z. B. "bus").
